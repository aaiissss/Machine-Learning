```python
#memanggil data set lewat gdrive
path = "/content/drive/MyDrive/Praktikum_ML/Praktikum03/Data/socr.csv"
```

```python
#membaca file csv menggunakan pandas
import pandas as pd

df = pd.read_csv(path)
display(df.head())
```

```python
df.describe()
```

```python
from copy import copy
df1 = (
    df[["Height(Inches)", "Weight(Pounds)"]]
    .rename(columns={"Height(Inches)": "tinggi_cm",
                     "Weight(Pounds)": "berat_kg"})
    .assign(tinggi_cm=lambda x: x["tinggi_cm"] * 2.54,
            berat_kg=lambda x: x["berat_kg"] * 0.453592)
    .round({"tinggi_cm": 2, "berat_kg": 2})
).copy()

df1.head()
```

```python
#membagi 80% data train dan 20% data test

from sklearn.model_selection import train_test_split

#variable independen / bebas

X = df1[["tinggi_cm"]]

#variable dependent / terikat
y = df1["berat_kg"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
    )

print('Jumlah data training' , len(X_train))
print('Jumlah data testing' , len(X_test))
```

6. pemodelan : training dan testing data[link text](https://)

```python
from sklearn.linear_model import LinearRegression

#buat objek instan dari class linearRegression bernama model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)

print("Koefisien (kg per cm):", model.coef_[0])
print("ntersep (kg):         ", model.intercept_)
print("R2 (test):                  ", r2)
print("MAE (test):                 ", mean_absolute_error(y_test, y_pred))
mse = mean_squared_error(y_test, y_pred) #default square=True
rmse = np.sqrt(mse)
print("RMSE (kg):                 ", rmse)
```

```python
slope = model.coef_[0]
intercept = model.intercept_

print(f"Persamaan: y = {slope:.3f}x + {intercept:.3f}")
```

```python
import matplotlib.pyplot as plt

#plot data scatter
plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, color='blue', label='Data Aktual')
# garis regresi
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Prediksi')
plt.xlabel('Tinggi Badan (cm)')
plt.ylabel('Berat Badan (kg)')
plt.title("REgresi Liear: Berat vs Tinggi")
plt.text(
    0.02, 0.98,
    f"y = {slope:.3f} * x + {intercept:.3f}\n$R^2$ = {r2:.3f}",
    transform=plt.gca().transAxes, va='top'
)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
y_pred_test = model.predict(X_test)

# Buat tabel hasil (tinggi, aktual, prediksi, dan error)
hasil = pd.DataFrame({
    "Tinggi (cm)": X_test["tinggi_cm"].to_numpy(),
    "Berat aktual (kg)": y_test.to_numpy(),
    "Berat Prediksi (kg)": y_pred_test,
})

# 1) Selisih error (positif = overpredict)
hasil["Selisih error (kg)"] = hasil["Berat Prediksi (kg)"] - hasil["Berat aktual (kg)"]

# 2) Akurasi per-baris (100 * (1 - |error|/aktual)), dibatasi 0-100
denom = hasil["Berat aktual (kg)"].replace(0, np.nan)  # antisipasi
hasil["Akurasi (%)"] = (1 - (hasil["Selisih error (kg)"].abs() / denom)).clip(lower=0, upper=1) * 100

hasil
```
